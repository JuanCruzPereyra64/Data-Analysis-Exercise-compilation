### Today, I am going to work with time series and aggregations.

#### We start by introducing the raw DataFrame. In this particular DataFrame, the dates have been uploaded as strings (plain text). The main objective for today is to transform those dates into real date types.

```python
import pandas as pd

raw_data = {
    'Order_ID': ['TX-001', 'TX-002', 'TX-003', 'TX-004', 'TX-005', 'TX-006', 'TX-007', 'TX-008'],
    'Date': ['2024-01-15', '2024-01-22', '2024-02-05', '2024-02-18', '2024-03-10', '2024-03-15', '2024-03-22', '2024-01-19'],
    'Category': ['Hardware', 'Peripherals', 'Hardware', 'Software', 'Software', 'Hardware', 'Peripherals', 'Hardware'],
    'Quantity': [2, 5, 1, 10, 4, 2, 3, 1],
    'Unit_Price': [350.0, 45.0, 800.0, 120.0, 150.0, 400.0, 60.0, 500.0]
}

df_sales = pd.DataFrame(raw_data)

print("Raw Sales Data:")
display(df_sales)
```

#### The first step will be to transform the dates in the 'Date' column to the correct variable type.

```python
print("Before the changes, the Date column variable type is: ")
print("\n")
df_sales.info()
print("\n")

df_sales['Date'] = pd.to_datetime(df_sales['Date'])

print("------------------")
print("\n")
print("After the changes it looks like this: ")
print("\n")
df_sales.info()
```

#### In the second step, I am going to add a new column to the DataFrame with the month of the sale. For this, I will extract the month of the year when the sale was made from the Date column.

```python
df_sales['Month'] = df_sales['Date'].dt.month_name()
print("Now we have the sales discriminated by month: ")
display(df_sales)
```

#### In the third step, I am going to add a new column calculating the total revenue.

```python
Total_revenue = df_sales['Quantity'] * df_sales['Unit_Price']
df_sales['Total_revenue'] = Total_revenue

print("Now we can see the total revenue per sale: ")
display(df_sales)
```

#### And last but not least, I am going to generate a new DataFrame that stores the aggregated information we extracted from the previous data.

```python
df_summary = df_sales.groupby(['Month', 'Category'])['Total_revenue'].sum().reset_index()

print("And now we have the final result, the summary DataFrame: ")
display(df_summary)
```
