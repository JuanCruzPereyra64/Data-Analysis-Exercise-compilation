### In today's exercise, I am going to work on trend analysis. For this case, we are going to simulate fulfillment data from a logistics center in this DataFrame.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Day 6: Trend Analysis - Cumulative Sums & Rolling Averages
# Simulating daily order fulfillment data for a logistics center
data = {
    'Date': pd.date_range(start='2024-05-01', periods=10, freq='D'),
    'Orders_Fulfilled': [120, 135, 110, 150, 165, 140, 180, 190, 175, 200]
}
df_logistics = pd.DataFrame(data)

print("Raw Logistics Data:")
display(df_logistics)
```

#### The first thing we are going to do is create a new column that calculates the cumulative sum of the orders fulfilled each day.

```python
df_logistics['Cumulative_Total'] = df_logistics['Orders_Fulfilled'].cumsum()

print("And now the new column looks like: ")
display(df_logistics)
```

#### The second step is to calculate the moving average and represent it in a new column. For this particular case, we are going to calculate the 3-day moving average.

```python
df_logistics['3_Day_MA'] = df_logistics['Orders_Fulfilled'].rolling(window=3).mean()

print("Dataframe with the 3 day moving average: ")
display(df_logistics)
```

#### Finally, we are going to visualize this information in a chart.

```python
df_logistics.plot(
    x='Date',
    y=['Orders_Fulfilled', '3_Day_MA'],
    figsize=(10,6),
    style=['o-', '--'],
    color=['#1f77b4', 'red']
)

plt.title('Trending of dispatch in the logistic center', fontsize= 14, pad= 15)
plt.xlabel('Date', fontsize= 12)
plt.ylabel('Order volume', fontsize= 12)
plt.legend(['Daily dispatchs' , 'Moving Average (3 Days)'])

plt.tight_layout()
plt.show()
```
