### In today's exercise, we are going to work with ABC Inventory Classification.

#### This is one of the most important analytics to review in the real world. It can be used in logistics, retail, sales, etc.

```python
import pandas as pd
import numpy as np

data = {
    'SKU': ['SKU-01', 'SKU-02', 'SKU-03', 'SKU-04', 'SKU-05', 'SKU-06', 'SKU-07', 'SKU-08'],
    'Description': ['Laptops', 'Mouse', 'Teclados', 'Monitores', 'Cables HDMI', 'Webcams', 'Auriculares', 'Sillas Ergonomicas'],
    'Unit_Cost': [800, 15, 30, 200, 5, 45, 60, 150],
    'Annual_Sold': [120, 500, 300, 80, 1000, 150, 200, 50]
}
df_inventory = pd.DataFrame(data)

print("Datos Crudos de Inventario:")
display(df_inventory)
```

#### We are going to classify the products in 3 groups:
- Class A: The small group of products that generates 80% of the revenue.
- Class B: Products with intermediate rotation and value.
- Class C: Products that have a low financial impact but occupy most of the physical space.

#### We are going to begin the analysis by creating a new column named Total_Value. This value is the multiplication of the units sold by their unit price.

```python
df_inventory['Total_Value'] = df_inventory['Unit_Cost'] * df_inventory['Annual_Sold']

print("The new column looks like: ")
display(df_inventory)
```

#### Now we are going to sort the products by their value

```python
df_inventory = df_inventory.sort_values(by='Total_Value', ascending= False).reset_index(drop=True)

print("And the dataframe looks like this: ")
display(df_inventory)
```

#### Now we are going to calculate what percentage of the whole warehouse each product represents.

```python
df_inventory['Cumulative_Value'] = df_inventory['Total_Value'].cumsum()
total_inventory_value = df_inventory['Total_Value'].sum()

df_inventory['Cumulative_Percentage'] = (df_inventory['Cumulative_Value'] / total_inventory_value) * 100

print("The percentage of each product is: ")
display(df_inventory)
```

#### Finally, it is time to classify every product into one of the 3 categories.

```python
conditions = [
    (df_inventory['Cumulative_Percentage'] <= 80),
    (df_inventory['Cumulative_Percentage'] <= 95)
]
choices = ['A', 'B']

df_inventory['ABC_Class'] = np.select(conditions, choices, default='C')

print("ABC Inventory Analysis Results: ")
display(df_inventory[['SKU', 'Description', 'Total_Value', 'Cumulative_Percentage', 'ABC_Class']])
```
