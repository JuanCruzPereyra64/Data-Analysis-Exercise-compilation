### In today's exercise, I am going to work with pivot tables.

#### These tables are mostly used in the corporate world to classify sales by region, month, or product.
#### They are an essential tool for business data analysis.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Day 5: Pivot Tables in Pandas
raw_data = {
    'Date': ['2024-01-10', '2024-01-15', '2024-02-20', '2024-02-25', '2024-03-05', '2024-03-12'],
    'Region': ['North', 'South', 'North', 'South', 'North', 'South'],
    'Category': ['Hardware', 'Hardware', 'Software', 'Software', 'Hardware', 'Software'],
    'Revenue': [1500, 2000, 1200, 1800, 1600, 2200]
}
df_sales = pd.DataFrame(raw_data)

print("Original Sales Data:")
display(df_sales)

# Creating the Pivot Table
# We want to see total Revenue grouped by Region (rows) and Category (columns)
pivot_df = pd.pivot_table(
    df_sales, 
    values='Revenue', 
    index='Region', 
    columns='Category', 
    aggfunc='sum',
    fill_value=0
)

print("\nBusiness Report: Pivot Table (Revenue by Region and Category):")
display(pivot_df)
```

#### As you can see, the result is a clean table that shows us how much revenue we have generated across every category and region.

```python
pivot_df.plot(kind='bar', figsize=(8, 6))

plt.title("Revenue per Region and Category", fontsize=14, pad=15)
plt.xlabel('Region', fontsize=12)
plt.ylabel('Revenue', fontsize=12)

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

#### Finally, we can generate a clear visualization that is easy for any stakeholder to understand.
