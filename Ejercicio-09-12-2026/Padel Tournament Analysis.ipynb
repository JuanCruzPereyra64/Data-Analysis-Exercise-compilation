## In this exercise, I am going to create a new dataframe containing null values, and then I will perform data cleaning on it. ##

```python
import pandas as pd
import numpy as np

# Raw tournament dataset

tournament = {
    'Player': ['Nico', 'Mati', 'Delfi', 'Lucas', 'Vale'],
    'Racket': ['Head One Ultralight', 'Babolat Technical', np.nan, 'Head One Ultralight', 'Bullpadel Hack'],
    'Unforced_Errors': [12, np.nan, 8, 15, 5],
    'Minutes_Played': [90, 85, 120, np.nan, 95]
}
df_tournament = pd.DataFrame(tournament)

display(df_tournament)
```

#### Let's start by checking how many null values we have to work with: ####

```python

print("Nulls to work on: ")
display(df_tournament.isna().sum())
```

### As we can see, we have: 1 null in the Racket column, 1 in Unforced_Errors, and 1 in Minutes_Played. ###

#### For this exercise, we are going to replace the null in Unforced_Errors with the mean (average). ####
#### We do this because replacing it with the mean won't significantly alter the overall statistical analysis of the dataset. ####

```python
# Filling the Unforced_Errors column with the median

print("Before changes: ")
display(df_tournament['Unforced_Errors']) 

mean_errors = df_tournament['Unforced_Errors'].mean()
df_tournament['Unforced_Errors'] = df_tournament['Unforced_Errors'].fillna(mean_errors)

print("After changes: ")
display(df_tournament['Unforced_Errors']) 
```

#### In the case of the null in the Racket column, we know this could be a missing data issue. ####
#### Therefore, I am going to replace the null with the 'Unknown' label to safely continue with the analysis. ####

```python
print("Befor the changes: ")
display(df_tournament['Racket'])

df_tournament['Racket'] = df_tournament['Racket'].fillna('Unknown')

print("After the changes: ")
display(df_tournament['Racket'])
```

#### We know that if a player has 0 or null minutes played in the tournament, it means the player did not participate. ####
#### So, for nulls in this specific column, I am going to drop the entire row. ####

```python
print("Before the changes: ")
display(df_tournament['Minutes_Played'])

df_tournament = df_tournament.dropna(subset=['Minutes_Played'])

print("After the changes")
display(df_tournament['Minutes_Played'])
```
